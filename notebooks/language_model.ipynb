```python
import sys
import os
sys.path.append(os.path.abspath('..'))

from quinex_application.language_model import LanguageModel
```

```python
llm = LanguageModel()
```

```python
models = llm.get_models()
print(models)
```

```python
from quinex_application.levenshtein import LevenshteinDistance
levenshtein = LevenshteinDistance()

from quinex_application.oeo_data import OEOData
oeo = OEOData("../data/oeo/etd.xlsx", "../data/oeo/oeo-full.owl", "../data/oeo/oeo.xlsx")
df_oeo = oeo.process_oeo_data()
```

```python
entity = "our work"
matches = levenshtein.find_top_matches(entity, df_oeo, top_n=10)
candidates = [match[0] for match in matches]
print(candidates)
```

```python
prompt = f"""You are given an entity mention and 10 candidate concepts. Your task is to choose the one that best matches the meaning of the entity mention.

Entity mention: "{entity}"

Candidates:
""" + "\n".join(f"{i+1}. {c}" for i, c in enumerate(candidates)) + f"""

Which of the candidates best matches the meaning of the entity "{entity}"? Only return the candidate text, nothing else.
If no candidate matches the meaning of the entity {entity} return None"""
print(prompt)
```

```python
response = llm.chat(message=prompt)
print(response)
```
