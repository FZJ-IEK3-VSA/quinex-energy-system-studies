##  **Import OEO**

```python
import sys
import os
sys.path.append(os.path.abspath('..'))

from quinex_application.oeo_data import OEOData

etd_path = "../data/raw/etd.xlsx"
ontology_path = "../data/raw/oeo-full.owl"
output_path = "../data/processed/oeo.xlsx"

oeo = OEOData(etd_path, ontology_path, output_path)
oeo.process_oeo_data()
```

```python
import pandas as pd
pd.set_option('display.max_colwidth', None)
oeo = pd.read_excel("../data/processed/oeo.xlsx")
display(oeo.head())
```

## **Levenshtein Distance**

```python
from quinex_application.levenshtein import LevenshteinDistance

levenshtein = LevenshteinDistance()

top_matches = levenshtein.find_top_matches("emission levels", oeo, levenshtein="extended_similarity")
for label, score in top_matches:
    print(f"🔍 Match: {label} — Score: {score:.2f}")
```

```python
match, score = levenshtein.get_best_match("emission levels", oeo, levenshtein="extended_similarity", score_limit=0.8)
print(f"🔍 Match: {match} — Score: {score:.2f}")
```
